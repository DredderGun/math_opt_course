Matplotlib - это библиотека для визуализации данных в Python. Она позволяет создавать широкий спектр графиков и диаграмм.

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
matplotlib.use('Agg') # other backends: TkAgg, wxAgg, Qt5Agg, Qt4Agg
params = {'legend.fontsize': 10,
          'legend.handlelength': 2,
          "axes.labelsize": 25,
          "xtick.labelsize": 10,
          "ytick.labelsize": 10,
          "lines.linewidth": 4,
           "axes.titlesize":15}
matplotlib.rcParams.update(params)
%matplotlib inline 
# plt.show() для отображения в окне
```

```python
x = np.linspace(1e-2, 10, 20)
fig, ax = plt.subplots()

q = x / 10.
ax.plot(x, (1 - np.sqrt(q)) / (1 + np.sqrt(q)), label='log(1/x)')
ax.set_yscale('log')
```

# Типы графиков

Гистограмма

```python
data = np.random.randn(10000)

fig, ax = plt.subplots()

ax.hist(data, bins=100, edgecolor='black')
ax.set_xlabel('Значение')
ax.set_ylabel('Частота')
fig.suptitle('Гистограмма случайных данных');
```

Ширину бинов можно регулировать

```python
xdata = np.array([1.2, 2.3, 3.3, 3.1, 1.7, 3.4, 2.1, 1.25, 1.3])
# xdata_gen = np.random.rand(100)*4 
xbins = np.array([1, 2, 4])

style = {'facecolor': 'none', 'edgecolor': 'C0', 'linewidth': 3}

fig, ax = plt.subplots()
ax.hist(xdata, bins=xbins, density=False, **style) # density=True замесо количества точек в промежутке показывает как плотно расположены точки в нем

ax.plot(xdata, 0*xdata, 'd')
ax.set_ylabel('Number per bin')
ax.set_xlabel('x bins (dx=1.0)');
```

Scatter plot

```python
fig, ax = plt.subplots()

x = np.random.rand(3)
y = np.random.rand(3)
colors = np.random.rand(3)
sizes = 1000 * np.random.rand(3)

print('x', x)
print('y', y)
print('colors', colors)
print('sizes', sizes)

ax.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='viridis')
ax.set_xlabel('x')
ax.set_ylabel('y')
fig.suptitle('Диаграмма разброса')

for i in range(3):
    plt.text(x[i], y[i], str(i))
```

# Линии уровней

Начертим линии уровней следующих функций
$$
\begin{gather}
f_1(x, y) = x^2 + y^2 \\
f_2(x, y) = \sin(x^2)e^{-\frac{x^2 + y^2}{5}} \\
f_3(x, y) = \begin{cases}
             \frac{x^2 y}{x^2 + y^2}, & (x, y) \neq (0, 0)\\
             0, & (x, y) = (0, 0)
            \end{cases}\\
f_4(x, y) = \sqrt[3]{x^3 + y^3}
\end{gather}
$$

```python
x, y = np.meshgrid(np.linspace(-1., 1., num=30), np.linspace(-1., 1., num=30))
x[x == 0] = 1e-5
y[y == 0] = 1e-5

def f_1(x, y):
    return x**2 + y**2

def f_2(x, y):
    return np.sin(x**2) * np.exp(-0.2 * (x**2 + y**2))

def f_3(x, y):
    result = x**2 * y / (x**2 + y**2)
    return result

def f_4(x, y):
    return np.cbrt(x**3 + y**3)

z_1 = f_1(x, y)
z_2 = f_2(x, y)
z_3 = f_3(x, y)
z_4 = f_4(x, y)

fig, ax = plt.subplots(2, 2)
ax[0, 0].contour(x, y, z_1, levels=np.logspace(-2., 2., num=30), cmap=plt.cm.rainbow_r)
ax[0, 1].contour(x, y, z_2, levels=np.logspace(-2., 2., num=30), cmap=plt.cm.rainbow_r)
ax[1, 0].contour(x, y, z_3, levels=np.logspace(-2., 2., num=30), cmap=plt.cm.rainbow_r)
ax[1, 1].contour(x, y, z_4, levels=np.logspace(-2., 2., num=30), cmap=plt.cm.rainbow_r);
```
